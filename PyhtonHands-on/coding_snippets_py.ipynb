```python
#finding minimum and maximum

arr = [1, 2, 3, 4, 5, 6, 7, 8, 9]

def find(arr):
  small, big = arr[0], arr[0]
  for i in arr[1:]:
    if i < small:
      small = i
    elif i > big:
      big = i
  return small, big
small, big = find(arr)
print("minimum element: ", small)
print("maximum element: ", big)

```

```python
#reverse a linked list

class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
node4 = Node(40)

node1.next = node2
node2.next = node3
node3.next = node4

#reversing
prev = None
current = node1
while current is not None:
  next_node = current.next
  current.next = prev     #flipping the pointer
  prev = current          #moving prev forward
  current = next_node     #moving current forward
head = prev       #prev is the new head

current = head
while current is not None:
  print(current.data, end="->")
  current = current.next
print("None")
```

```python
#finding the middle node

class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
node4 = Node(40)
node5 = Node(50)

node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node5

slow = node1
fast = node1
while fast is not None and fast.next is not None:
  slow = slow.next
  fast = fast.next.next
print("Middle node: ", slow.data)
```

```python
#reversing a stack

class Stack:
  def __init__(self):
    self.values = []
  def push(self, x):
    self.values = [x] + self.values
  def pop(self):
    return self.values.pop(0)
  def reverse(self):
    self.values = self.values[::-1]

s = Stack()
s.push(10)
s.push(20)
s.push(30)
s.push(40)
print("before reversing: ", s.values)
s.reverse()
print("after reversing: ", s.values)
```

```python
#binary search

arr = [10, 20, 30, 40, 50]
target = 40

def binary(arr, target):
  left, right = 0, len(arr) - 1
  while left <= right:
    mid = (left + right) // 2
    if arr[mid] == target:
      return mid
    elif arr[mid] < target:
      left = mid + 1
    else:
      right = mid - 1
  return -1
print(binary(arr, target))
```

```python
#First occurrence of a target in a sorted array

arr = [10, 20, 30, 30, 30, 40, 50]
target = 30

def first_occurrence(arr, target):
  left, right = 0, len(arr) - 1
  result = -1
  while left <= right:
    mid = (left + right) // 2
    if arr[mid] == target:
      result = mid
      right = mid - 1
    elif arr[mid] < target:
      left = mid + 1
    else:
      right = mid - 1
  return result

print(first_occurrence(arr, target))
```

```python
#insertion sort

def insertion(arr):
  n = len(arr)

  for i in range (1, n):
    key = arr[i]
    j = i - 1
    while j >= 0 and key < arr[j]:
      arr[j + 1] = arr[j]
      j -= 1
    arr[j + 1] = key

arr = [40, 30, 20, 10]
insertion(arr)
print("Sorted array is: ", arr)
```

```python
#merge sort

arr = [60, 20, 40, 10, 90, 30, 80, 50]

def merge_sort(arr):
  if len(arr) <= 1:
    return arr
  mid = len(arr) // 2
  l_half = arr[:mid]     #left part
  r_half = arr[mid:]     #right part
  l_half = merge_sort(l_half)
  r_half = merge_sort(r_half)
  return merge(l_half, r_half)

def merge(left, right):
  new = []
  i, j = 0, 0

  while i < len(left) and j < len(right):
    if left[i] < right[j]:
      new.append(left[i])    #adding element from left to new list
      i += 1
    else:
      new.append(right[j])    #adding element from right to new list
      j += 1

  new.extend(left[i:])
  new.extend(right[j:])
  return new

print("Sorted array is:", merge_sort(arr))
```

```python
#quick sort

arr = [5, 8, 1, 2, 6, 3, 9]

def quick_sort(arr, low, high):
  if low < high:
    pivot = partition(arr, low, high)
    quick_sort(arr, low, pivot-1)
    quick_sort(arr, pivot+1, high)

def partition(arr, low, high):
  p = arr[low]
  i = low + 1
  j = high
  while True:
    while i <= j and arr[i] <= p:     #left to right loop -> searches for elements greater than pivot element
      i += 1
    while i <= j and arr[j] >= p:     #right to left loop -> searches for elements lesser than pivot element
      j -= 1
    if i <= j:
      arr[i], arr[j] = arr[j], arr[i]   #swapping the elements
    else:
      break
  arr[low], arr[j] = arr[j], arr[low]
  return j

quick_sort(arr, 0, len(arr) - 1)
print("Sorted array is:", arr)
```

```python
#binary search using recursion

arr = [10, 20, 30, 40, 50]
target = 30

def binary(arr, target, left, right):
  #BASE CASE
  if left > right:
    return -1

  mid = (left + right) // 2
  if arr[mid] == target:
    return mid
  elif arr[mid] < target:
    return binary(arr, target, mid+1, right)
  else:
    return binary(arr, target, left, mid-1)

print(binary(arr, target, 0, len(arr)-1))
```

```python
# finding kth largest element using heap

import heapq

arr = list(map(int, input("Enter elements: ").split()))
k = int(input("Enter a value for k: "))

def kth_largest(arr, k):
  heap = []
  for i in arr:
    heapq.heappush(heap, i) #adding the elements to the heap one by one
    if len(heap) > k:
      heapq.heappop(heap)
  return heap[0]

result = kth_largest(arr, k)
print("The kth largest element is: ", result)
```
